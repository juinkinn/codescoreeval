````python
import json
import os
````

## Load data

````python
subset_path = "./data/subset_100_2.jsonl"
metadata_path = "./data/metadata.jsonl"
output_path = "./data/annotated_subset_2.jsonl"

# load metadata
metadata = {}
with open(metadata_path, "r", encoding="utf-8") as f:
    for line in f:
        item = json.loads(line)
        metadata[item["id"]] = item

# load subset
data = []
with open(subset_path, "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))
````

## Sequential annotation

**INPUT:**

- At first, enter starting index you want to annotate.
- In the loop, if you want to stop, type ```close```.

````python
view_path = "annotation_view.txt"

annotated = []
done_ids = set()

if os.path.exists(output_path):
    with open(output_path, "r", encoding="utf-8") as f:
        for line in f:
            item = json.loads(line)
            annotated.append(item)
            done_ids.add(item["sub_id"])

print("Loaded", len(data), "samples")
print("Already annotated:", len(done_ids))

start_index = input("Start from index (enter = auto): ")

if start_index == "":
    start_index = None
else:
    start_index = int(start_index)


def extract_problem_id(sub_id):
    return sub_id.rsplit("_", 1)[0]


total = len(data)

for i, row in enumerate(data):

    if start_index is not None:
        if i < start_index:
            continue
    else:
        if row["sub_id"] in done_ids:
            continue

    prob_id = extract_problem_id(row["sub_id"])
    meta = metadata.get(prob_id, {})

    description = meta.get("description", "")
    time_limit = meta.get("time_limit", "")
    memory_limit = meta.get("memory_limit", "")
    title = meta.get("title", "")

    with open(view_path, "w", encoding="utf-8") as f:

        f.write(f"SAMPLE {i+1}/{total}\n")
        f.write("="*80 + "\n")
        f.write(f"Problem: {title}\n")
        f.write(f"Problem ID: {prob_id}\n")
        f.write(f"Submission: {row['sub_id']}\n")
        f.write("="*80 + "\n\n")

        f.write("DESCRIPTION\n\n")
        f.write(description + "\n\n")

        f.write("LIMITS\n")
        f.write(f"Time: {time_limit}\n")
        f.write(f"Memory: {memory_limit}\n\n")

        f.write("SUBMISSION INFO\n")
        f.write(f"Language: {row['lang']}\n")
        f.write(f"Correct: {row['correct']}\n")
        f.write(f"Runnable: {row['runnable']}\n")
        f.write(f"Exec time: {row['exec_time_ms']}\n")
        f.write(f"Memory: {row['memory_mb']}\n\n")

        f.write("CODE\n")
        f.write("-"*80 + "\n")
        f.write(row["code"] + "\n")
        f.write("-"*80 + "\n")

    print("\n-----------------------------------")
    print(f"Progress: {i+1}/{total}")
    print(f"Remaining: {total - i - 1}")
    print(f"View sample at: {view_path}")
    print("-----------------------------------")

    print("Commands: score (0-5) | skip | close")

    correctness = input("correctness_score: ")

    if correctness.lower() == "close":
        break
    if correctness.lower() == "skip":
        continue

    efficiency = input("efficiency_score: ")

    if efficiency.lower() == "close":
        break

    syntax = input("syntax_score: ")

    if syntax.lower() == "close":
        break

    row["subset_index"] = i
    row["correctness_score"] = correctness
    row["efficiency_score"] = efficiency
    row["syntax_score"] = syntax

    annotated = [x for x in annotated if x["sub_id"] != row["sub_id"]]

    annotated.append(row)
    
    annotated_sorted = sorted(annotated, key=lambda x: x["subset_index"])

    with open(output_path, "w", encoding="utf-8") as f:
        for item in annotated_sorted:
            f.write(json.dumps(item, ensure_ascii=False) + "\n")

print("\nAnnotation stopped.")
print("Total annotated:", len(annotated))
````

## Update a row

````python
edit_index = int(input("Enter subset index to edit: "))

row = data[edit_index]

prob_id = extract_problem_id(row["sub_id"])
meta = metadata.get(prob_id, {})

description = meta.get("description", "")
time_limit = meta.get("time_limit", "")
memory_limit = meta.get("memory_limit", "")
title = meta.get("title", "")

with open(view_path, "w", encoding="utf-8") as f:

    f.write(f"EDIT SAMPLE {edit_index}/{len(data)}\n")
    f.write("="*80 + "\n")
    f.write(f"Problem: {title}\n")
    f.write(f"Problem ID: {prob_id}\n")
    f.write(f"Submission: {row['sub_id']}\n")
    f.write("="*80 + "\n\n")

    f.write("DESCRIPTION\n\n")
    f.write(description + "\n\n")

    f.write("LIMITS\n")
    f.write(f"Time: {time_limit}\n")
    f.write(f"Memory: {memory_limit}\n\n")

    f.write("SUBMISSION INFO\n")
    f.write(f"Language: {row['lang']}\n")
    f.write(f"Correct: {row['correct']}\n")
    f.write(f"Runnable: {row['runnable']}\n")
    f.write(f"Exec time: {row['exec_time_ms']}\n")
    f.write(f"Memory: {row['memory_mb']}\n\n")

    f.write("CODE\n")
    f.write("-"*80 + "\n")
    f.write(row["code"] + "\n")
    f.write("-"*80 + "\n")

print("Sample written to:", view_path)


annotations = []
if os.path.exists(output_path):
    with open(output_path, "r", encoding="utf-8") as f:
        for line in f:
            annotations.append(json.loads(line))


target = None
for item in annotations:
    if item["sub_id"] == row["sub_id"]:
        target = item
        break


if target is not None:
    print("Current scores:")
    print("correctness:", target["correctness_score"])
    print("efficiency:", target["efficiency_score"])
    print("syntax:", target["syntax_score"])
else:
    print("Sample này chưa annotate, sẽ tạo mới.")


new_correct = input("New correctness_score: ")
new_eff = input("New efficiency_score: ")
new_syn = input("New syntax_score: ")


annotations = [x for x in annotations if x["sub_id"] != row["sub_id"]]


new_item = {
    "subset_index": edit_index,
    "sub_id": row["sub_id"],
    "correctness_score": new_correct,
    "efficiency_score": new_eff,
    "syntax_score": new_syn
}

annotations.append(new_item)

annotations = sorted(annotations, key=lambda x: x["subset_index"])


with open(output_path, "w", encoding="utf-8") as f:
    for item in annotations:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")

print("Score updated.")
````
