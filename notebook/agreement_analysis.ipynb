```python
import pandas as pd
import json
from scipy.stats import spearmanr
from sklearn.metrics import cohen_kappa_score
```

```python
INPUT_DIR = '../data'
TYPE = 'syntax'
annotated_1 = pd.read_json(f"{INPUT_DIR}/annotated_subset_1.jsonl", lines=True)
annotated_2 = pd.read_json(f"{INPUT_DIR}/annotated_subset_2.jsonl", lines=True)
syntax_1 = pd.read_json(f'{INPUT_DIR}/new_prompt/{TYPE}_1.json', lines=True)
syntax_2 = pd.read_json(f'{INPUT_DIR}/new_prompt/{TYPE}_2.json', lines=True)
corr = pd.read_json(f'{INPUT_DIR}/old_prompt/correctness.json')

annotated = pd.concat([annotated_1, annotated_2])
syntax = pd.concat([syntax_1, syntax_2])
```

```python
merged_1 = pd.merge(
    annotated_1[['sub_id', f'{TYPE}_score']],
    syntax_1[['sub_id', f'{TYPE}_score']],
    on='sub_id',
    suffixes=('_human', "_llm")
)
kappa = cohen_kappa_score(
    merged_1[f'{TYPE}_score_human'],
    merged_1[f'{TYPE}_score_llm'],
    weights='quadratic'
)


rho, p_value = spearmanr(merged_1[f'{TYPE}_score_human'], merged_1[f'{TYPE}_score_llm'])
merged_1['diff'] = abs(merged_1[f'{TYPE}_score_human'] - merged_1[f'{TYPE}_score_llm'])
agreement_count = (merged_1['diff'] == 0).sum()

print(f'Weighted kappa: {kappa:.3f}')
print(f"Spearman ρ = {rho:.3f}, p = {p_value:.3f}")
print(f'Agreement count: {agreement_count}')
cols = ['sub_id', f'{TYPE}_score_human', f'{TYPE}_score_llm']
merged_1[merged_1[f'{TYPE}_score_human'] != merged_1[f'{TYPE}_score_llm']][cols].to_json('../data/syn_diff_100_1.jsonl', orient='records', lines=True, force_ascii=False)
```

```python
merged_2 = pd.merge(
    annotated_2[['sub_id', f'{TYPE}_score']],
    syntax_2[['sub_id', f'{TYPE}_score']],
    on='sub_id',
    suffixes=('_human', "_llm")
)
kappa = cohen_kappa_score(
    merged_2[f'{TYPE}_score_human'],
    merged_2[f'{TYPE}_score_llm'],
    weights='quadratic'
)


rho, p_value = spearmanr(merged_2[f'{TYPE}_score_human'], merged_2[f'{TYPE}_score_llm'])
merged_2['diff'] = abs(merged_2[f'{TYPE}_score_human'] - merged_2[f'{TYPE}_score_llm'])
agreement_count = (merged_2['diff'] == 0).sum()

print(f'Weighted kappa: {kappa:.3f}')
print(f"Spearman ρ = {rho:.3f}, p = {p_value:.3f}")
print(f'Agreement count: {agreement_count}')
cols = ['sub_id', f'{TYPE}_score_human', f'{TYPE}_score_llm']
merged_2[merged_2[f'{TYPE}_score_human'] != merged_2[f'{TYPE}_score_llm']][cols].to_json('../data/syn_diff_100_2.jsonl', orient='records', lines=True, force_ascii=False)
```
