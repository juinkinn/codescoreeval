```python
import json
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data = []

with open('../../data/submissions.jsonl', "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))
```

```python
print(data[0].keys())
```

```python


df = pd.DataFrame(data)

sampled, _ = train_test_split(
    df,
    train_size=100,
    stratify=df["correct"],
    random_state=42
)

sampled_data = sampled.to_dict("records")
```

```python
sampled_data
```

```python
part1 = sampled_data[:50]
part2 = sampled_data[50:]

with open("../../data/subset_100_1.jsonl", "w", encoding="utf-8") as f:
    for row in part1:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

with open("../../data/subset_100_2.jsonl", "w", encoding="utf-8") as f:
    for row in part2:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")
```
