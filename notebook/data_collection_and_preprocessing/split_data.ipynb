```python
from datasets import load_dataset, Dataset
import pandas as pd
```

```python
df = pd.read_json("/kaggle/input/datasets/trungkinnguynhong/code-contests/final_filtered.jsonl", lines=True)

# Convert to HF Dataset
ds = Dataset.from_pandas(df)
```

```python
META_COLS = [
    'source', 'id', 'title', 'description',
    'time_limit', 'memory_limit'
]

meta_df = df[META_COLS]

meta_dataset = Dataset.from_pandas(meta_df)
meta_dataset.to_json("metadata.jsonl", orient="records", lines=True)
print("Saved metadata.jsonl")
```

```python
rows = []

for row in df.to_dict(orient="records"):
    pid = row["id"]

    # correct submissions
    for idx, sub in enumerate(row.get("correct_submissions", [])):
        rows.append({
            "sub_id": f"{pid}_c{idx}",
            "code": sub["code"],
            "lang": sub.get("language", sub.get("lang")),
            "correct": 1
        })

    # incorrect submissions
    for idx, sub in enumerate(row.get("incorrect_submissions", [])):
        rows.append({
            "sub_id": f"{pid}_i{idx}",
            "code": sub["code"],
            "lang": sub.get("language", sub.get("lang")),
            "correct": 0
        })

sub_df = pd.DataFrame(rows)
sub_ds = Dataset.from_pandas(sub_df)

sub_ds.to_json("submissions.jsonl", orient="records", lines=True)
print("Saved submissions.jsonl")
```
