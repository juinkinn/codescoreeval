```python
import json

sub_map = {}

with open("../../data/submissions.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        item = json.loads(line)
        sub_map[item["sub_id"]] = item
```

```python
RENAME_MAP = {
    "correctness_score": "gt_correctness_score",
    "efficiency_score": "gt_efficiency_score",
    "readability_score": "gt_readability_score"
}

def rename_fields(input_path, output_path):
    with open(input_path, "r", encoding="utf-8") as fin, \
         open(output_path, "w", encoding="utf-8") as fout:

        for line in fin:
            item = json.loads(line)

            for old_key, new_key in RENAME_MAP.items():
                if old_key in item:
                    item[new_key] = item.pop(old_key)

            fout.write(json.dumps(item, ensure_ascii=False) + "\n")
```

```python
rename_fields("../../data/train.jsonl", "../../data/train_gt.jsonl")
rename_fields("../../data/test.jsonl", "../../data/test_gt.jsonl")
```

```python
def append_fields(input_path, output_path, sub_map):
    with open(input_path, "r", encoding="utf-8") as fin, \
         open(output_path, "w", encoding="utf-8") as fout:

        for line in fin:
            item = json.loads(line)
            sub_id = item["sub_id"]

            if sub_id in sub_map:
                full_info = sub_map[sub_id]
                
                merged = {**full_info, **item}
            else:
                print(f"Warning: sub_id {sub_id} not found")
                merged = item

            fout.write(json.dumps(merged, ensure_ascii=False) + "\n")
```

```python
append_fields("../../data/train_gt.jsonl", "../../data/submissions_train.jsonl", sub_map)
append_fields("../../data/test_gt.jsonl", "../../data/submissions_test.jsonl", sub_map)
```
