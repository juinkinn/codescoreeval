```python
import json
import random
from collections import defaultdict
```

```python
random.seed()

test_data = []
with open("../../data/original_test.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        test_data.append(json.loads(line))

problem_groups = defaultdict(list)

for row in test_data:
    problem_groups[row["id"]].append(row)

print(f"Total problems: {len(problem_groups)}")

selected_problem_ids = random.sample(list(problem_groups.keys()), 5)

print("Selected problems:")
for pid in selected_problem_ids:
    print(pid, len(problem_groups[pid]))

subset = []

for pid in selected_problem_ids:
    subset.extend(problem_groups[pid])

print(f"Total submissions: {len(subset)}")
```

```python

random.shuffle(subset)

mid = len(subset) // 2

subset1 = subset[:mid]
subset2 = subset[mid:]

with open("../../data/human_annotated_test_set/subset_1.jsonl", "w", encoding="utf-8") as f:
    for row in subset1:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

with open("../../data/human_annotated_test_set/subset_2.jsonl", "w", encoding="utf-8") as f:
    for row in subset2:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print(len(subset1), len(subset2))
```
