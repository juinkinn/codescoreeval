```python
import json
import os
```

```python
subset_path = "../data/human_annotated_test_set/subset_1.jsonl"
metadata_path = "../data/metadata.jsonl"
submission_path = "../data/submissions.jsonl"

output_path = "../data/human_annotated_test_set/annotated_subset_1.jsonl"
view_path = "annotation_view.txt"

metadata = {}

with open(metadata_path, "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        metadata[obj["id"]] = obj

submissions = {}

with open(submission_path, "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)
        submissions[obj["sub_id"]] = obj

data = []

with open(subset_path, "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))
```

```python
annotated = []
done_ids = set()

if os.path.exists(output_path):
    with open(output_path, "r", encoding="utf-8") as f:
        for line in f:
            item = json.loads(line)
            annotated.append(item)
            done_ids.add(item["sub_id"])

print("Loaded:", len(data))
print("Already annotated:", len(done_ids))

start = input("Start index (Enter = auto): ")

if start == "":
    start = None
else:
    start = int(start)

total = len(data)


for i, row in enumerate(data):

    if start is not None:
        if i < start:
            continue
    else:
        if row["sub_id"] in done_ids:
            continue

    problem = metadata[row["id"]]
    submission = submissions[row["sub_id"]]

    with open(view_path, "w", encoding="utf-8") as f:

        f.write(f"SAMPLE {i+1}/{total}\n")
        f.write("=" * 80 + "\n")
        f.write(f"Problem: {problem['title']}\n")
        f.write(f"Problem ID: {row['id']}\n")
        f.write(f"Submission: {row['sub_id']}\n")
        f.write("=" * 80 + "\n\n")

        f.write("DESCRIPTION\n\n")
        f.write(problem["description"])
        f.write("\n\n")

        f.write("CODE\n")
        f.write("-" * 80 + "\n")
        f.write(submission["code"])
        f.write("\n")
        f.write("-" * 80 + "\n")

    print("\n--------------------------------")
    print(f"{i+1}/{total}")
    print(view_path)
    print("--------------------------------")

    print("Commands: score | skip | close")

    correctness = input("Correctness (0-5): ")

    if correctness.lower() == "close":
        break

    if correctness.lower() == "skip":
        continue

    efficiency = input("Efficiency (0-5): ")

    if efficiency.lower() == "close":
        break

    readability = input("Readability (0-5): ")

    if readability.lower() == "close":
        break

    result = dict(row)

    result["subset_index"] = i
    result["human_correctness_score"] = int(correctness)
    result["human_efficiency_score"] = int(efficiency)
    result["human_readability_score"] = int(readability)

    annotated = [x for x in annotated if x["sub_id"] != row["sub_id"]]
    annotated.append(result)

    annotated.sort(key=lambda x: x["subset_index"])

    with open(output_path, "w", encoding="utf-8") as f:
        for item in annotated:
            f.write(json.dumps(item, ensure_ascii=False) + "\n")

print("\nAnnotation stopped.")
print("Total annotated:", len(annotated))
```
