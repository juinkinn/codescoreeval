```python
import json
from collections import defaultdict
from sklearn.metrics import cohen_kappa_score

annotation_file = "../data/human_annotated_test_set/annotated_set.jsonl"

with open(annotation_file, "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

metrics = [
    ("correctness_score", "human_correctness_score"),
    ("efficiency_score", "human_efficiency_score"),
    ("readability_score", "human_readability_score"),
]

print(f"Number of samples: {len(data)}")

for gt_col, human_col in metrics:
    gt = [int(x[gt_col]) for x in data]
    human = [int(x[human_col]) for x in data]

    kappa = cohen_kappa_score(gt, human, weights="quadratic")

    mismatches = []
    for item in data:
        g = int(item[gt_col])
        h = int(item[human_col])
        if g != h:
            mismatches.append(
                (item["sub_id"], g, h, abs(g - h))
            )

    mismatches.sort(key=lambda x: (-x[3], x[0]))

    groups = defaultdict(list)
    for x in mismatches:
        groups[x[3]].append(x)

    print("\n" + "=" * 100)
    print(f"{gt_col.upper()}")
    print("=" * 100)
    print(f"Kappa     : {kappa:.4f}")
    # print(f"Mismatches: {len(mismatches)}/{len(data)}")

    # if not mismatches:
    #     continue

    # for diff in sorted(groups.keys(), reverse=True):
    #     print(f"\nDiff = {diff} ({len(groups[diff])} samples)")
    #     print("-" * 100)

    #     for sub_id, g, h, _ in groups[diff]:
    #         print(f"{sub_id:20} GT={g}   Human={h}")
```
